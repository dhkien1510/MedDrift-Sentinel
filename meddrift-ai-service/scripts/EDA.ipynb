```python
from datasets import load_dataset
import matplotlib.pyplot as plt
import seaborn as sns


ds = load_dataset("flaviagiammarino/vqa-rad", split="train")
df = ds.to_pandas()

df.head(5)
```

```python

df_question = df[['question']]

print(df_question)
```

```python
import pandas as pd
text_len = [len(text) for text in df_question['question']]

text_len.sort()

df = pd.DataFrame(text_len, columns=['length'])

plt.figure(figsize=(10, 6))
# bins=30 chia dữ liệu thành 30 cột nhỏ để dễ nhìn
plt.hist(df['length'], bins=30, edgecolor='black')
plt.xlabel('Độ dài văn bản')
plt.ylabel('Số lượng câu hỏi')
plt.title('Phân phối độ dài câu hỏi')
plt.show()
```
